# Семинар 5. Анализ рынка и конкурентной среды

Рынок - это все потенциальные клиенты нашего продукта

#### Структура рынка

* PAM - потенциально доступный объем рынка
  * TAM - ёмкость целевого рынка
    * SAM - доступный объем обслуживаемого рынка
      * SOM - достижимый объем рынка

**Potential Available Market** 

Это все люди/организации, которые могут стать вашими клиентами сейчас и в обозримом будущем. 

Обзоры и тренды: Bloomberg, Forbes. Открытые источники: Открытые данные всемирного банка, База данных ООН Comtrade, HubSpot, Statista и другие.

**Total Addrressable Market**

Это клиенты, имеющие потребность, но не обязательно готовые платить за ваше или аналогичное решение. 

Statista, Росстат, Налоговая, раздел статистики, 2gis, Ведомственная статистика, ВЦИОМ, результаты опросов населения и другие

**Serviceable Available Market**

Часть целевого рынка, которая уже пользуется вашим продуктом и продуктами конкурентов. Здесь проводим анализ конкурентов и соотносим доли рынка со своими.

Отчёты публичных компаний, SensorTower, Similarweb, Google Trends, Crunchbase и другие

**Serviceable & Obtainable Market**

Это да доля от доступного рынка, которую вы можете захватить в соответствии со своими "мощностями" - объемом производимой продукции или оказываемых услуг, бюджетами на ФОТ, рекламу и другие

SensorTower, Similarweb, Google Trends, Crunchbase, свои исследования и другие

#### Анализ конкурентов

**Прямые** - компании, которые предлагают аналогичные или очень похожие продукты или услуги вашей компании.

**Непрямые** - компании или продукты, которые могут не предлагать аналогичные продукты или услуги, но имеют сопоставимые характеристики или выполняют аналогчиные функции, что делает их вторичными конкурентами

**Косвенные (заменители)** - компании или продукты, которые предлагают альтернативные способы удовлетворения тех же потребностей у потребителя, по сравлению с вашим продуктом

#### Определяем структуру рынка

* Выделяем основных игроков и их доли
* Ищем конкурентов, похожих на нас бюджетом м ресурсами
* Определяем доли рынка
* Исследуем изменения

#### SWOT-анализ

**Сильные стороны (S)** 
* крупная база клиентов
* высокий процент лояльных
* разнообразный ассортимент
* готовая инфраструктура для развития (склады, транспорт и так далее)

**Слабые стороны (W)** 
* низкая узнаваемость бренда
* высокая долговая нагрузка
* старый ремент

**Возможности (O)** 
* расширение ассортимента услуг компании
* расширение мощностей
* внедрение маркетинга

**Угрозы (T)** 
* появилось большое количество более крутых сетевых компаний
