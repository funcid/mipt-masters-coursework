Скопируйте описание параметров установленного вами дистрибутива в текстовый файл.

### Решение 

Терминал не открывается... Пошел на StackOverflow: https://askubuntu.com/questions/1435918/terminal-not-opening-on-ubuntu-22-04-on-virtual-box-7-0-0

Пересоздал опять, воспользовавшись советом из https://askubuntu.com/a/1437432

Выполню команды:

```bash
uname -a
echo $SHELL
echo $XDG_CURRENT_DESKTOP
ps -p 1 -o comm=
cat /proc/cmdline
ps -p 2 -o pid,comm,cmd
systemctl list-units --type=service --state=running
cat /etc/os-release
```
