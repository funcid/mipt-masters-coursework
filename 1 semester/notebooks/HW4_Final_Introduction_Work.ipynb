## Задание 2

Объясните, что делает команда: `ps -aux | grep root | wc -l >> root`

* Примечание: Если вы встречаете неизвестную команду Linux либо неизвестные параметры команды, можете вызвать встроенную помощь: `man <команда>`
* Пример: `man ps`, `man grep` `man wc`

### Решение 

![man](./man.png)

* `ps -aux`

Показывает список всех процессов, запущенных в системе, с информацией о пользователях, PID, потреблении памяти/CPU и т.д.

* `| grep root`

Фильтрует вывод, оставляя только строки, где встречается слово `root`.

* `| wc -l`

Считает количество строк, переданных на вход.
То есть — количество процессов, где упоминается `root`.

* `>> root`

Добавляет (`>>`) результат (число) в конец файла с именем `root`.
Если файла `root` нет — он будет создан.
Если есть — число просто добавится в конец.
