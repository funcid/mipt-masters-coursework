# Семинар 1. Введение в продуктовую аналитику. Работа в команде аналитики. Основные типы бизнес-метрик

**Продукт** это цифровое решение, разрабатываемое и улучшаемое на основе данных о поведении пользователей и бизнес-целях.

**Непрерывный цикл улучшений**: сбор данных, анализ, доработка, запуск новой версии.

Поведение, потребности и обратная связь пользователей напрямую влияют на функциональность и дизайн продукта.

Продукт создает устойчивую ценность как для пользователей, так и для бизнеса.

**Продуктовый аналитик** специалист, который собирает, анаоизирует и интерпретирует данные о поведении пользователей и эффективности продукта, чтобы помогать принимать обоснованные решения по его развитию

1. *Сбор и обеспечение качественных о поведении пользователей в продукте*
2. Проведение юзабилити-тестов с живыми пользователями
3. *Анализ данных для выявления узких мест и формулировка гипотез*
4. Разработка архитектуры базы данных для хранения логов
5. *Создание и поддержка дашбордов для коммуникации инсайтов с командой*
6. Ведение переговоров с инвесторами о привлечении финансирования 

Как продуктовый аналитик влиеяет на решения в бизнесе?

Продуктовый аналитик превращает сырые данные в понятные инсайты: он выявляет, что работает, а что - нет, оценивает потенциальный эффект от изменений, помогает расставлять приоритеты на основе метрик и обеспечивает фактологическую основу для стратегических и тактических решений команды продукта.
