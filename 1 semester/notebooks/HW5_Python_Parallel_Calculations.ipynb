```python
import threading
import time
```

## Задача 1

Напишите программу, которая создаёт 2 потока для вычисления квадратов и кубов целых чисел от 1 до 10.

```python
def print_squares():
    for i in range(1, 11):
        print(f"{i}^2 = {i ** 2}")

def print_cubes():
    for i in range(1, 11):
        print(f"{i}^3 = {i ** 3}")

t1 = threading.Thread(target=print_squares)
t2 = threading.Thread(target=print_cubes)

t1.start()
t2.start()

t1.join()
t2.join()
```

## Задача 2

Напишите программу, которая создаёт несколько потоков для выполнения функции, которая выводит целые числа от 1 до 10 с задержкой в 1 секунду.

```python
thread_count = 2

def print_numbers(thread_name):
    for i in range(1, 11):
        nanos = time.time_ns()  
        print(f"{thread_name}: {i} (время: {nanos} нс); ")
        time.sleep(1)

threads = []
for n in range(thread_count):
    t = threading.Thread(target=print_numbers, args=(f"Поток-{n+1}",))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```
