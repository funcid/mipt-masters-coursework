#### Зачем бизнесу аналитика?

Аналитика помогает понять, что происходить, почему, что делать дальше и какой будет эффект?

#### Web-аналитика

**Определение:** системный сбор, обработка и интерпретация данных о поведении пользователей на сайтах и в приложениях, чтобы принимать продуктовые и маркетинговые решения и измерять их эффект.

#### Где нужна Web-аналитика

* Госсектор и НКО
* E-commerce и маркетплейсы
* Медиа/контент/SEO-проекты

* **Цель бизнеса** - прибыль: без нее он не работает
* **Прибыль = доход** - расходы: на нее влияет маржинальность
* **Базовые драйверы:** число продаж, конверсия, средний чек
* Все начинается с низкоуровневых метрик: привлечения

#### Виды сайтов

* Сайты-визитки (цель - познакомить, KPI: сессии, отказы, клики)
* Лидогенераторы (цель - оставить контакт, KPI: CR валидных лидов)
* Интернет-магазин (цель - покупка, KPI: AOV, CR в оплату, повторные заказы)

### Яндекс Метрика

1. Визит (сессия) - период активности одного пользователя на сайте
2. Внутренний и внешний переход
3. Глубина просмотра - количество просмотренных страниц за визит
4. Время на сайте / на странице - разница между метками времени соседних хитов; последняя страница визита может "не иметь времени", если не было следующего события
5. Цель и достижение цели
6. Источник трафика и метки
7. Отказ (bounce)
8. Роботы (боты)

Очень интересный инструмент от Яндекс - https://datalens.ru/

#### ClientID .vs UserID

* ClientID - анонимный ID браузера/устройства, приявязанный к Coockie ID
* UserID - постоянный ID клиента
