Указываем интересующую нас папку и берем из нее все .IPYNB файлы

```python
import glob
import os
import subprocess
from concurrent.futures import ThreadPoolExecutor, as_completed

# Настраиваем пути
path = [
    '1 семестр',
    'notebooks/',
]
mask = os.path.join(os.getcwd(), *path, '*.ipynb')
files = glob.glob(mask)

print(f'Файлов затронуто: {len(files)}')
```

Конвенртируем все файлы в .PDF

```python
def convert_to_pdf(file_path):
    try:
        # Папка ../misc/pdf относительно текущего файла
        src_dir = os.path.dirname(file_path)
        dst_dir = os.path.normpath(os.path.join(src_dir, "../misc/pdf"))
        os.makedirs(dst_dir, exist_ok=True)

        # Имя выходного файла
        base_name = os.path.splitext(os.path.basename(file_path))[0]
        output_file = os.path.join(dst_dir, f"{base_name}")

        # Конвертация
        subprocess.run(
            [
                "jupyter", "nbconvert",
                "--log-level", "ERROR",
                "--to", "webpdf",
                "--allow-chromium-download",
                "--output", output_file,
                file_path
            ],
            check=True,
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL,
        )
        return f"✅ {os.path.basename(file_path)} → {output_file}"
    except subprocess.CalledProcessError:
        return f"❌ Ошибка при конвертации {os.path.basename(file_path)}"
```

```python
max_workers = min(4, os.cpu_count() or 2)
with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = [executor.submit(convert_to_pdf, f) for f in files]
    for future in as_completed(futures):
        print(future.result())
```
