# Семинар 2. Unit-экономика

![Метрики продукта](./Метрики%20продукта.png)

$$Метрики \space продукта \times новых \space пользователей = метрики \space роcта$$

### Метрики продукта

* Retention (удержание)
* LTV (Life Time Value, сколько клиент принесет за все время в среднем)
* Time spend (среднее потраченное время клиентом за все время)
* SSR (процент обращения в поддержку, без привлечения оператора)
* Customer Acquisition Cost (САС, стоимость привлечения клиента)

### Метрики роста

* DAU (Daily Active Users, New Users $\times$ Retention)
* Прибыль (New Users $\times$ LTV)
* Запущенные сценарии (New Users $\times$ Coversation into running scenario)
* Paying users

### AAARRR

AAARRR (Pirate Metrics) — фреймворк из шести ключевых этапов воронки продукта:

A — Awareness — узнаваемость: сколько людей узнали о продукте

A — Acquisition — привлечение: сколько пользователей пришли

A — Activation — активация: сколько получили первую ценность (например, сделали регистрацию или первую покупку)

R — Retention — удержание: сколько пользователей возвращаются

R — Revenue — доход: сколько пользователей приносят деньги

R — Referral — рекомендации: сколько пользователей приводят других

### NSM

NSM (North Star Metric) — "Северная звезда" продукта — главная метрика, отражающая основную ценность, которую продукт приносит пользователям, и напрямую связана с долгосрочным ростом бизнеса.

Примеры:

Spotify → время прослушивания музыки

Airbnb → количество забронированных ночей

Slack → число активных сообщений в командах

eCommerce → количество оформленных заказов или AOV $\times$ Retention

Зачем нужно:
NSM помогает команде фокусироваться не на куче метрик, а на одной — той, которая лучше всего отражает "здоровье" и развитие продукта.

![NSM](./NSM.png)

### Основные метрики

* ARPPU — средняя выручка на одного платящего пользователя
* C1 — конверсия в первую покупку
* ARPU — средняя выручка на одного пользователя (включая неплатящих)
* CPU — стоимость привлечения одного пользователя
* CAC — стоимость привлечения клиента (платящего)
* COGS — себестоимость проданных товаров или услуг
* APC — число покупок на одного платящего пользователя
* AOV — средний чек
