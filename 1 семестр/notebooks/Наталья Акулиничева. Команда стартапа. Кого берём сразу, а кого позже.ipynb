# Наталья Акулиничева. Команда стартапа: кого берём сразу, а кого позже

CEO Logic BPM. Lemma Group

##### В чем самая большая проблема при найме сотрудников в стартап?

* Продавец воздуха 
* Нет денег

##### Что же делать? Как найти тех, кто вам подходит?

* Продавать ценности - найти правльных по ценностям людей - самое главное в подборе. Как это определить на интервью?
* Продавать энергию - без энергии в стартапе невозможно
* Продавать идею - вы должны четко понимать, за счет чего будете выигрывать на рынке

##### Где и как искать?

* Каналы в Телеграм - самый большой отклик + бесплатно
* Объявление должно соответствовать требованиям канала - посмотрите, как написано у других
* Добавьте в объявление искренности - откликнется больше людей
* Не собеседуйте каждого соискателя - вы выгорите. Сформируйте 3-5 вопросов по каждой специальности, которые можно отправить в качестве первой проверки.

##### Структура стартапа

* Не ищите универсальных специалистов - это маловероятно
* Нарисуйте четкую структуру компании и опишите необходимые роли
* Сократите все процессы и роли, без которых можно обойтись
* Если есть партнер, договоритесь обо всем на берегу

##### Если фаундер - продакт

* Обычно хорошо знает бизнесовую часть продукта
* Имеет доступ к будущим клиентам
* Легко превратить продукт в кастомную разработку для фаундера

##### Когда в стартапе должен появиться маркетинг?

Сразу. Клиентский опыт после продажи - ваше отличие на рынке, поможет при сборе инвестиций.

##### Сделать команду

1. Безопасная среда - даем право на ошибку
2. Ошибки разбираем по методу "черного ящика"
3. Формируем среду быстрой проверки гипотез - ищем Product-Market Fit
4. Метрики - оцифровываем результаты всех процессов
5. Убираем бюрократию и контроль
6. Проводим регулярные встречи с сотрудниками - формируем общее видение

* Пять пороков команды
* Цель. Процесс непрерывного совершенствования
* Результат чужими руками. Путеводитель для руководителей
