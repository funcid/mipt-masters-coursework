* Аналитика

## Методы проверки гипотез

* Экспертные интервью
* Глубинные интервью - общение один на один, которое позволяет собрать достаточно данных (опросив 20-100 человек) для решения определенной задачи. Главные правила: открытые вопросы, акцент на прошлом, не нарывайтесь на комплимент, подтверждайте прошлое/намерения, люди врут и они не эксперты
* Опросы - устный или письменный сбор данных с помощью заранее сформулированных вопросов для решения определенных задач

## Как применять RAT?

* Выписать все рискованные предположения для будущего продукта и бизнеса, которые мы видим на текущий момент
* Отранжировать от наиболее рискованных до наименее рискованных
* Пойти проверять топ рискованных предположений

## Проверка гипотез решения 

* UX тесты
  * Формируем пользовательскую задачу
  * Даем её пользователю
  * Смотрим, как он её решает
  * Фиксируем проблемы / вопросы / неожиданные действия
  * Валидируем обратную связь и радуемся инсайтам
* Прототипирование
  * Эскиз кривой рукой
  * Красивые маркеры
  * Интерактивные прототипы
  * Рабочие прототипы созданные с помощью no-code / нейронок
* Понятное описание задачи
* MVP (Minimum viable product)
* Гэмба
* Фокус-группа
* Этнография
* Коридорные исследования

## MVP / MMP / MPL

* Minimum viable product - продавать стыдно
* Minimum marketable product - моожно стыдно
* Minimum lovable product - нравится пользоваться, может быть < MVP
