# Антон Шишков. Основы Apache Airflow

Инженер данных, Сбер (AmazMe)

Apache Airflow нужен для автоматизации, планирования и мониторинга сложных рабочих процессов (workflows), особенно в области обработки данных (ETL/ELT), машинного обучения и отчетности; он позволяет определять задачи как Python-код в виде DAG (направленных ациклических графов), управлять зависимостями между ними и визуально отслеживать их выполнение. Это мощный инструмент для инженеров данных и аналитиков, позволяющий строить надежные и масштабируемые конвейеры данных. 

### Для кого?

**Data engineer** - потоки обработки данных

**BI** - обеспечить регулярные отчеты

**Feature Engineering** - настроить потоки создания новых признаков при построении модели

1. Airflow предназначен чисто для Batch нагрузки (вместо Streaming)
2. Оркестратор задач - не является ETL инструментом, планировщиком

### Архитектура Airflow

1. Scheduler
2. DAG Processor
3. Webserver
4. DAG Folder
5. Metadata DB

Опциональные компоненты:

Worker, Triggerer, Plugins

### Локальное развертывание

Airflow живет только в Linux

### Что такое DAG?

DAG (направленный ациклический граф) — это основная концепция Airflow, объединяющая задачи и организующая их с помощью зависимостей и отношений, указывающих, как они должны выполняться.
