### 1. Потребительские сегменты (ПС)

- Нишевый рынок (меньше конкуренция, фокус, выше LTV)
- Массовый рынок (большой рынок, накомпления, низкий LTV)
- Дробное сегментирование (меньше клиентов, разные маркетинговые стратегии, LTV выше)
- Многопрофильные предприятия, многосторонние платформы (не на старте)

Лучшая стратегрия: старт с нишевого рынка и переход к дробному сегментированию

### 2. Ценостные предложения (ЦП)

- Новизна - **частично**, виден тренд среди молодых людей на систематизацию жизни и геймификацию, продукт может забрать финансовую плоскость
- Производительность - **нет**, есть уже готовые разрозненные решения
- Изготовление на заказ - **да**, личные финансовые рекомендации основанные на портфеле клиента
- "Делать свою работу" - **частично**, конкуренты (банки) интегрируются между собой и вытесняют малый fintech
- Дизайн - **нет**, современные российские продукты уже показывают конкурентный дизайн
- Бренд/статус - **нет**, конкуренты недостижимы
- Цена - **частично**, рынок до конца не сформирован, реальные финансовые консультанты дороги
- Уменьшение расходов - **да**, не нужно создавать excel табличку или руками что-либо вводить
- Снижение риска - **нет**, услуги конкурируют на том же уровне рисков
- Доступность - **да**, финансовые рекомендации доступны за 3-5к рублей или при премиальном пакете услуг, а ИИ вариант дешевле на порядок
- Удобство/применимость - **да**, удобство агрегации всех активов
