## Книга "Построение бизнес-моделей": FuncFund

Авторы: Александр Остервальдер и Ив Пинье

## Шаблон

### Девять структурных блоков

1. Потребительские сегменты (ПС)
2. Ценостные предложения (ЦП)
3. Каналы сбыта (КС)
4. Взаимоотношения с клиентами (ВК)
5. Потоки поступления доходов (ПД)
6. Ключевые ресурсы (КР)
7. Ключевые виды деятельности (КД)
8. Ключевые партнеры (КП)
9. Структура издержек (СИ)

### 1. Потребительские сегменты (ПС)

- Нишевый рынок (меньше конкуренция, фокус, выше LTV)
- Массовый рынок (большой рынок, накомпления, низкий LTV)
- Дробное сегментирование (меньше клиентов, разные маркетинговые стратегии, LTV выше)
- Многопрофильные предприятия, многосторонние платформы (не на старте)

Лучшая стратегрия: старт с нишевого рынка и переход к дробному сегментированию

### 2. Ценостные предложения (ЦП)

- Новизна - **частично**, виден тренд среди молодых людей на систематизацию жизни и геймификацию, продукт может забрать финансовую плоскость
- Производительность - **нет**, есть уже готовые разрозненные решения
- Изготовление на заказ - **да**, личные финансовые рекомендации основанные на портфеле клиента
- "Делать свою работу" - **частично**, конкуренты (банки) интегрируются между собой и вытесняют малый fintech
- Дизайн - **нет**, современные российские продукты уже показывают конкурентный дизайн
- Бренд/статус - **нет**, конкуренты недостижимы
- Цена - **частично**, рынок до конца не сформирован, реальные финансовые консультанты дороги
- Уменьшение расходов - **да**, не нужно создавать excel табличку или руками что-либо вводить
- Снижение риска - **нет**, услуги конкурируют на том же уровне рисков
- Доступность - **да**, финансовые рекомендации доступны за 3-5к рублей или при премиальном пакете услуг, а ИИ вариант дешевле на порядок
- Удобство/применимость - **да**, удобство агрегации всех активов

### 3. Каналы сбыта (КС)

Какие каналы взаимодействия были бы желательны для ваших потребительских сегментов? Как мы взаимодействуем с ними сейчас? Как связаны наши каналы между собой? Какие из них наиболее эффективны? Какие более выгодны?

Прямые:
- Торговые агенты
- Продажи через Интернет

Непрямые:
- Фирменные магазины
- Партнерские магазины
- Оптовики

**Этапы:**

1. Информационный - как мы повышаем осведомленность потребителя о товарах и услугах нашей компании?
2. Оценочный - как мы помогаем клиентам оценивать ценностные предложения нашей компании?
3. Продажный - каким образом мы предоставляем клиентам возможность приобретения определенных товаров и услуг?
4. Доставка - как мы доставляем клиентам наши ценностные предложения?
5. Постпродажный - как мы обеспечиваем клиентам постпродажное обслуживание?

### 4. Взаимоотношения с клиентами (ВК)

Отношения какого типа ждет каждый потребительский сегмент? Какие отношения установлены? Каких расходов они требуют? Как они интегрированы в общую схему бизнес-модели?

1. Персональная поддержка - **нет**
2. Особая персональная поддержка - **нет**
3. Самообслуживание - **да**, клиент получает понятные услуги и свободно ими оперирует
4. Автоматизированное обслуживание - **да**, частный случай самооьслуживания
5. Сообщества - **возможно**, тяжело развивать при ограниченности ресурсов
6. Совместное создание - **возможно**, однако это сложная модель и требует энтузиазма

### 5. Потоки поступления доходов (ПД)

За что клиенты действительно готовы платить? За что они платят в настоящее время? Каким образом платят? Как они предпочли бы платить? Какую часть общей прибыли приносит каждый поток поступления доходов?

1. Продажа активов - **нет**
2. Плата за использование - **да**, каждая финансовая рекомендация стоит денег
3. Оплата подписки - **да**, расширенная финансовая аналитика
4. Аренда/рента/лизинг - **нет**
5. Лицензии - **нет**
6. Брокерские проценты - **нет**
7. Реклама - **да**, размещение баннеров с описаниями услуг банков, экосистем, брокеров

### 6. Ключевые ресурсы (КР)

Какие ключевые ресурсы нужны для наших ценностных предложений? Наших каналов сбыта? Взаимоотношений с клиентами? Потоков доходов?

1. Материальные ресурсы - сервера, исходный код
2. Интеллектуальные ресурсы - архитектура, древо продуктовых метрик, домены
3. Персонал - разработчики, аналитики, тестировщики, менеджер, продажники и маркетологи
4. Финансы - кредитные линии, средства компании

### 7. Ключевые виды деятельности (КД)

Каких видов деятельности требуют наши ценностные предложения? Наши каналы сбыта? Наши взаимоотношения с клиентами? Наши потоки поступления доходов?

1. Производство - **нет**
2. Разрешение проблем - **да**, помогает клиентам копить средства и балансировать широкий портфель без большого объема знаний
3. Платформы/сети - **возможно**, предоставление API для клиентов по агрегации активов и B2B сегмент помощи банкам для коммуникации с Premium, Private слоями

### 8. Ключевые партнеры (КП)

Виды партнерских отношений: стратегическое сотрудничество между неконкурирующими компаниями, соконкуренция: стратегическое партнерство между конкурентами, совместрые предприятия для запуска новых бизнес-проектов, отношения производителя с поставщиками для гарантии получения качественных комплектующих.

Кто является нашими ключевыми партнерами? Кто наши основные поставщики? Какие ключевые ресурсы мы получаем от партнеров? Какой ключевой деятельностью занимаются наши партнеры?

1. Оптимизация и экономия в сфере производства - **нет**
2. Снижение риска и неопределенности - **нет**
3. Поставки ресурсов и совместная деятельность - **да**, банки предоставляющие данные клиентов

### 9. Структура издержек (СИ)

Какие наиболее важные расходы предполагает наша бизнес-модель? Какие из ключевых ресурсов наиболее дороги? Какие ключевые виды деятельности требуют наибольших затрат?

1. Ориентация на издержки - **частично**, минимальное кол-во вычислительных ресурсов, доступные LLM
2. Ориентация на ценность - **нет**, рекомендации все соизмеримы в рамках их стоимости, если заплатить в 2 раза за токен, результат не будет лучше в 2 раза
3. Фиксированные издержки - **да**, ФОТ
4. Переменные издержки - **да**, вычислительные ресурсы, маркетинг
5. Экономия на масштабе - **нет**
6. Эффект диверсификации - **нет**

## Стили

### Разделение бизнес-модели

#### Стиль #1

Концепция "разделение" предполагает наличие трех фундаментальных форм бизнес-деятельности:

1. бизнес, ориентированный на клиентов
2. бизнес, ориентированный на инновации
3. бизнес, ориентированный на инфраструктуру

У каждой формы свои правила

Три основные формы физнеса

| Сфера        | Инновационные продукты                                                                 | Управление взаимоотношениями с клиентами                                              | Управление инфраструктурой                                                                 |
|--------------|------------------------------------------------------------------------------------------|----------------------------------------------------------------------------------------|---------------------------------------------------------------------------------------------|
| **Экономика** | Ранний выход на рынок позволяет ставить выгодные цены; главное — скорость.              | Высокие затраты на привлечение клиентов требуют высокой доли расходов потребителей; главное — экономия на масштабе. | Высокая доля фиксированных затрат требует больших объемов производства; главное — экономия на масштабе. |
| **Культура**  | Битва за таланты; низкий порог входа; множество мелких игроков.                         | Битва за ассортимент; доминируют немногие крупные компании.                            | Битва за масштаб; консолидация; доминируют немногие крупные компании.                      |
| **Конкуренция** | Ориентация на сотрудников; повышенное внимание к творческим личностям.                | Ориентация на услуги; мышление «клиент всегда прав».                                   | Ориентация на издержки; упор на стандартизацию, предсказуемость и эффективность.           |

### Длинный хвост 

#### Стиль #2

Длинный хвост (Long Tail) - это модель спроса, при которой совокупность множества редких, нишевых товаров даёт больший общий объём продаж, чем узкая группа хитов.

### Многосторонние платформы

#### Стиль #3

Многосторонняя платформа - это бизнес-модель, которая соединяет две и более взаимозависимые группы участников и создаёт ценность именно благодаря их взаимодействию.

Классика: маркетплейсы, соцсети, платежные системы, Uber-подобные сервисы. Они не «продают товары», они создают среду, где продавцы находят покупателей, водители - пассажиров, авторы - читателей, а котики - лайки.

Ключевая фишка - сетевые эффекты: чем больше участников одной группы, тем ценнее платформа для другой. Именно поэтому такие проекты либо взлетают, либо стремительно превращаются в пустыню.

### FREE как бизнес-модель

#### Стиль #4

FREE как бизнес-модель - это подход, при котором основная часть продукта или услуги предоставляется пользователю бесплатно, а доход компания получает за счёт дополнительных механизмов монетизации.

Суть в том, что «бесплатно» - это не благотворительность, а способ резко расширить аудиторию и затем монетизировать лишь небольшую её часть. Работает там, где себестоимость цифрового копирования стремится к нулю.

Freemium - базовый функционал бесплатно, расширенные возможности за оплату.

Реклама - пользователи ничего не платят, платят рекламодатели.

Субсидирование одной стороны - одна группа пользователей получает сервис бесплатно, другая оплачивает (классика для платформ).

Раздача продукта ради продаж дополняющих товаров - «бритва бесплатно, лезвия платные».

### Открытые бизнес-модели

#### Стиль #5

Открытые бизнес-модели - это подход к созданию и использованию ценности, при котором компания сознательно использует внешние идеи, технологии, ресурсы или каналы, а также сама открывает свои разработки наружу. Главное - не вариться в собственном соку, а объединять внутренние и внешние источники инноваций.

В основе - сотрудничество, обмен знаниями и совместное создание ценности.
Это может работать в двух направлениях:

Входящая открытость (outside-in) - компания привлекает внешние идеи, технологии, исследования, стартапы, open-source-решения, чтобы ускорить инновации и снизить затраты.

Исходящая открытость (inside-out) - компания предоставляет свои разработки, API, данные, лицензии или платформы другим игрокам, чтобы расширить рынок или создать экосистему.

Типичные примеры: open-source проекты, открытые API (как у Stripe или Spotify), корпоративные акселераторы, совместные R&D, лицензирование технологий.

| Принципы инноваций | Закрытые инновации | Открытые инновации |
|--------------------|--------------------|---------------------|
| **Таланты** | Талантливые люди в нашей области работают на нас. | Мы должны работать с талантливыми людьми как внутри компании, так и со стороны. |
| **Источник ценности НИОКР** | Чтобы получать прибыль от НИОКР, мы должны сами делать открытия, разрабатывать их и применять. | Внешние НИОКР тоже могут создавать ценность, и чтобы претендовать на неё, мы сами тоже должны заниматься исследованиями. |
| **Путь на рынок** | Если мы сделаем открытие сами, то выйдем на рынок первыми. | Не обязательно самим проводить исследования, чтобы получать от них прибыль. |
| **Идеи и конкурентное преимущество** | Если большинство лучших идей в отрасли будут нашими, мы выиграем. | Мы выиграем, если сможем наилучшим образом использовать как свои, так и чужие идеи. |
| **Контроль инноваций** | Мы должны контролировать инновационный процесс, чтобы конкуренты не воспользовались нашими идеями. | Мы должны получать прибыль от использования нашими инноваций другими, и приобретать чужие инновации, если это выгодно и соответствует нашим интересам. |


FuncFund лучше подходит модель открытых инноваций,
потому что финтех - это гонка скорости, интеграций и постоянного эксперимента, а не медитативное изобретение велосипеда в одиночестве.

### Генерация идей

1. Создание команды: подбор участников с разным опытом; распределение ролей (фасилитатор, таймкипер, фиксатор); договорённость о правилах обсуждения.
2. Погружение: формулировка задачи; сбор ограничений и контекста; ознакомление с существующими решениями; определение рамок для генерации.
3. Расширение: мозговые штурмы и другие креативные техники; сбор всех идей без фильтрации; комбинирование и варьирование идей.
4. Выбор критериев: определение параметров оценки; расстановка приоритетов; фиксация критериев отбора.
5. Прототипирование: выбор нескольких идей; быстрые MVP/черновые решения; проверка ключевых гипотез.

### Визуализация

Визуализация: создание наглядных материалов, которые упрощают восприятие ключевых элементов бизнес-плана; использование схем, диаграмм, графиков и инфографики для отражения структуры продукта, бизнес-модели, рынка, финансов и дорожной карты; оформление макетов, пользовательских сценариев и визуальных концепций; приведение данных и идей в форму, которую можно быстро понять, оценить и презентовать.

### Прототипирование

Прототипирование: создание упрощённых моделей ключевых решений бизнес-плана; разработка тестовых макетов продукта, сервисных сценариев или функциональных черновиков; проверка основных гипотез через быстрые эксперименты; получение ранней обратной связи для корректировки концепции до перехода к полноценной разработке или внедрению.

### Сторитейлинг

Сторитейлинг: формирование ясного и увлекательного повествования, которое объясняет идею бизнес-плана через понятную историю; создание логичной последовательности - от проблемы и героя (клиента) до решения и ценности продукта; использование метафор, примеров и сценариев, которые помогают удержать внимание аудитории; подача информации так, чтобы проект воспринимался живым, осмысленным и легко запоминаемым.

### Сценарии

Сценарии: создание детальных вариантов развития событий, связанных с использованием продукта или реализацией бизнес-плана; описание шагов пользователя, клиента или партнёра в разных ситуациях; моделирование типичных, стрессовых и альтернативных сценариев поведения рынка или системы; проработка последовательности действий, точек взаимодействия и возможных результатов, чтобы заранее увидеть сильные и слабые стороны концепции.
