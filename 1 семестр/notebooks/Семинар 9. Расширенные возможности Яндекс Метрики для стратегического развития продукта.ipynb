# Семинар 9. Расширенные возможности Яндекс Метрики для стратегического развития продукта

Google Ads. Ремаркетинг Google Ads

Organic Search. Переход из органического потока - последний непрямой канал

#### First Interaction

Модель ранжирования, где вероятность клика зависит только от качества конкретного документа для конкретного пользователя. Каждая встреча “с чистого листа”, без учёта того, что было раньше.

#### Time Decay

Метод, где более свежие данные получают больший вес, а старые - постепенно «умирают». Отлично подходит, когда пользовательские предпочтения меняются быстрее, чем мемы в Telegram.

#### Position Based

Вероятность клика = качество документа × фиксированный коэффициент позиции. То есть часть кликов — за контент, часть — за то, что документ «сидит повыше».

| Метод                        | Когда подходит                                                                 | Ключевая причина                                                               |
| ---------------------------- | ------------------------------------------------------------------------------ | ------------------------------------------------------------------------------ |
| **First Interaction**        | Длинные воронки, важный этап «знакомства», сложные продукты                    | Нужно понять, какой источник впервые привёл пользователя                       |
| **Time Decay**               | Динамичные рынки, быстро меняющееся поведение, короткие циклы принятия решения | Свежие события важнее старых                                                   |
| **Position Based (U-Shape)** | И длинные, и средние воронки, где важны и первый, и последний шаг              | Балансирует между открытием и закрытием сделки                                 |
| **Last Non-Direct Click**    | Когда брендовые переходы и прямые визиты искажают картину эффективности        | Надо отделить реальный вклад рекламы от «пришёл напрямую, потому что уже знал» |
