# Семинар 10. Основы аналитики мобильного приложения на базе AppMetrica

Сколько в среднем шагов делает пользователь в приложении, чтобы добраться до экрана?

А сколько хочется?

## Диплинк

Диплинк (Deep Link) - это ссылка, ведущая прямо в нужную точку мобильного приложения

* Конкретный товар
* Открытый чат
* Конкретная статься или акция

### Типы диплинков (Regular)

Работают если приложение установленно:

* Ретаргенинг
* Push-кампании
* Email-рассылка

### Типы диплинков (Deferred)

Отложенные диплинки, сценарий:

1. Пользователь кликает -> идет в App Store / Google Play
2. Устанавливает приложение
3. При первом запуске попадает прямо на нужный экран

Regular считаются устаревшими.

### Умные ссылки (Smart Link)

1. Определяет платформу (iOS, Android), установлено ли приложение
2. Если установленно - обычный deeplink
3. Если нет - в магазин + отложенный deeplink после установки

## Постбэки

Постбэк (Post Back) - это сигнал партнеру, что пользователь выполнил целевое действие с целью атрибуции трафика.

### Системы
