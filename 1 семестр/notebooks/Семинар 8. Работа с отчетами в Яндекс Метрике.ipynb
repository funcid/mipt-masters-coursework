# Семинар 8. Работа с отчетами в Яндекс Метрике

### Зачем нам Яндекс Метрика?

##### Понимать поведение пользователей на сайте 

С метрикой видим: какая страница самая популярная; с какой страницы чаще всего уходят; как ведут себя мобильные пользователи и desktop

##### Оценивать эффективность рекламы и каналов трафика

Метрика нужна не "для отчетности", а чтобы **понимать, окупается ли реклама**.

##### Находить и исправлять проблемы сайта

##### Принимать решения на основе данных, а не интуиции

### Что такое цель в Метрике?

Цель = важное действие пользователя (заявка, покупка, звонок/клик по телефону, подписка/регистрация)

##### Зачем нужны цели?

Считать конверсии; сравнивать источники трафика; оценивать эффективность рекламы;

Без целей видим только посещаемость
С целями видим, что приночит результат

### Автоматические цели в Яндекс Метрике

Это цели, которые Метрика создает сама, без ручной настройки.

**Плюсы:**
1. Быстрый старт, даже если нет специалиста
2. Можно увидеть первые конверсии без сложной настройки
3. Помогают не забыть про важные действия

**Минусы:**
1. Могут собирать лишние действия (поиск по сайту)
2. Не всегда совпадают с реальными целями бизнеса
3. Если полагаться только на автоцели, можно искажать аналитику (конверсий много, а заявок мало)

### Обзор отчетов в Яндекс Метрике

**Сводка** - главный экран для ежедневного контроля

##### Популярные группы отчетов

1. Посещаемость - сколько людей, когда приходят, как меняется трафик
2. Источники - Сводка (откуда приходят пользователи, какие каналы приводят конверсии)
3. UTM-метки - детальный анализ рекламных компаний по тегам
4. Аудитория и поведение - география, устройства, новые/вернувшиеся, глубина просмотра и др.

### Роботность: кто такие "роботы" в Метрике?

**Роботы** - это программы, а не живые пользователи. Программы, которые посещают сайты с какими-то определенными целями.

Бывают:

1. Поисковые (индексируют сайт: Яндекс, Google, ...)
2. Сервисные (проверка ссылок, мониторинг доступности)
3. Вредоностные/скликивающие

Роботы искажают статистику: завышают количество визитов, портят показатели отказов, глубины, времени на сайте. 

### Семплирование: когда Метрика показывает не все данные

**Семплирование** - это когда Метрика считает отчет не по всем визитам, а по части данных.

Зачем?

1. Ускорение выгрузки отчетов, когда данных очень много
2. Включается при объемах более 500k~ визитов за период
3. Позволяет построить отчет, который иначе был бы слишком тяжелым

Особенность:

1. Результаты отчетов при семлировании - это оценка, а не точные цифры
2. Чем меньше доля выборки, тем ниже точность

### Сегментация трафика: зачем делить пользователей на группы?

Общий отчет != ответы на все вопросы

В одном отчете "все визиты подряд" смешиваются: новые и вернувшиеся, мобильные и desktop, с заявкой и без.

### Вебвизор: "видеозапись" визитов пользователей

Записывает поведение пользователя на сайте в формате "видео": прокрутки, клики, движения мыши, ввод формы

## Ключевые идеи

1. Цель позволяет измерять конверсии
2. Отчеты и Сводка дают картину по трафику и заявкам
3. Роботность и семплирование влияют на качество данных
4. Сегментация помогает выделять нужную аудиторию
5. Вебвизоры и карты показывают, что улучшить на сайте
