## Цель

Закрепить навыки работы с файлами в Python

```python
def resolve_path(path):
    return f"../misc/Файлы к домашнему заданию 2/{path}"
```

## Задание 1

Копирование содержимого одного файла в другой.
Создайте программу, которая копирует содержимое файла `source.txt` в новый файл `destination.txt`.

### Решение 

```python
# Копирование содержимого из source.txt в destination.txt
with open(resolve_path("source.txt"), "r", encoding="utf-8") as src:
    content = src.read()

with open(resolve_path("destination.txt"), "w", encoding="utf-8") as dst:
    dst.write(content)

print("Файл успешно скопирован.")
```

## Задание 2

Подсчёт стоимости заказа из файла.
Напишите программу, которая считывает файл `prices.txt`, содержащий информацию о товарах: название, количество и цену, и подсчитывает общую стоимость заказа.

### Решение 

```python
total_cost = 0

with open(resolve_path("prices.txt"), "r", encoding="utf-8") as file:
    for line in file:
        name, quantity, price = line.split()
        total_cost += int(quantity) * float(price)

print(f"Общая стоимость заказа: {total_cost}")
```

## Задание 3

Подсчёт количества слов в файле.
Напишите программу, которая подсчитывает количество слов в текстовом файле `text_file.txt` и выводит результат на экран.

### Решение 

```python
with open(resolve_path("text_file.txt"), "r", encoding="utf-8") as file:
    text = file.read()

words = text.split()
print(f"Количество слов в файле: {len(words)}")
```

## Задание 4

Копирование уникального содержимого одного файла в другой.
Создайте программу, которая считывает строки из файла `input.txt`, сохраняет только уникальные строки и записывает их в новый файл `unique_output.txt`.

### Решение 

```python
with open(resolve_path("input.txt"), "r", encoding="utf-8") as infile:
    unique_lines = set(infile.readlines())

with open(resolve_path("unique_output.txt"), "w", encoding="utf-8") as outfile:
    outfile.writelines(unique_lines)

print("Уникальные строки записаны в unique_output.txt")
```
