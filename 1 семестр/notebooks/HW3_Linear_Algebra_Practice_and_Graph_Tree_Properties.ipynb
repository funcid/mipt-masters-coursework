## Задание 1

Найдите все решения линейных систем: 

a)
$
\begin{cases}
y + 3z = -1 \\
2x + 3y + 5z = 3 \\
3x + 5y + 7z = 6
\end{cases}
\quad\quad$
б)
$
\begin{cases}
x + 2y + 3z = -4 \\
2x + 3y + 4z = 1 \\
3x + 4y + 5z = 6
\end{cases}
$

### Решение 

**a)** Из первого уравнения $y+3z=-1\Rightarrow y=-1-3z$. Подставим в второе:

$$
2x+3(-1-3z)+5z=3\Rightarrow 2x-3-9z+5z=3\Rightarrow 2x-4z=6\Rightarrow x=3+2z.
$$

Подставляем в третье:

$$
3(3+2z)+5(-1-3z)+7z=6\Rightarrow 9+6z-5-15z+7z=6\Rightarrow 4-2z=6\Rightarrow z=-1.
$$

Тогда $x=3+2(-1)=1; y=-1-3(-1)=2$.


Ответ: $(x,y,z)=(1,2,-1)$

---

**б)** Вычтем первое уравнение из второго: $(2)-(1)\Rightarrow x+y+z=5$. То же получаем и из $(3)-(2)$, значит одно из трёх уравнений лишнее. Решаем систему

$$
\begin{cases}
x+2y+3z=-4,\\
x+y+z=5.
\end{cases}
$$

Вычитая второе из первого: $y+2z=-9\Rightarrow y=-9-2z$. Подставляем в $x+y+z=5$:

$$
x-9-2z+z=5\Rightarrow x=14+z.
$$

Пусть параметр $t=z$. Тогда

$$
(x,y,z) = (14+t,-9-2t,t) = (14,-9,0)+t(1,-2,1); t\in\mathbb{R}.
$$

Ответ: бесконечно много решений

## Задание 2

Найдите координаты вектора в базисе $e_1 = (2, 3)^{\mathrm{T}}, \quad e_2 = (3, 4)^{\mathrm{T}}$, если известны его координаты $(v'_1, v'_2)$ в базисе $e'_1 = (1, -1)^{\mathrm{T}}, \quad e'_2 = (2, -3)^{\mathrm{T}}$.

### Решение 

Запишем вектор через координаты в новом базисе:

$$
\mathbf{a} = v'_1 e'_1 + v'_2 e'_2
= v'_1 \begin{pmatrix} 1 \ -1 \end{pmatrix} + v'_2 \begin{pmatrix} 2 \ -3 \end{pmatrix}
= \begin{pmatrix} v'_1 + 2v'_2 \ -v'_1 - 3v'_2 \end{pmatrix}.
$$

То же самое в старом базисе:

$$
\mathbf{a} = v_1 e_1 + v_2 e_2
= v_1 \begin{pmatrix} 2 \\
3 \end{pmatrix} + v_2 \begin{pmatrix} 3 \\ 
4 \end{pmatrix} = \begin{pmatrix} 2v_1 + 3v_2 \\ 
3v_1 + 4v_2 \end{pmatrix}.
$$

Приравняем компоненты (ведь это один и тот же вектор!):

$$
\begin{cases}
2v_1 + 3v_2 = v'_1 + 2v'_2,\\
3v_1 + 4v_2 = -v'_1 - 3v'_2.
\end{cases}
$$

Решаем систему относительно $(v_1, v_2)$.

Умножим первое уравнение на $3$ и второе на $-2$, чтобы исключить $v_1$:

$$
\begin{cases}
6v_1 + 9v_2 = 3v'_1 + 6v'_2,\\
-6v_1 - 8v_2 = 2v'_1 + 6v'_2.
\end{cases}
$$

Складываем: $v_2 = 5v'_1 + 12v'_2.$

Теперь подставляем в первое уравнение:

$$
2v_1 + 3(5v'_1 + 12v'_2) = v'_1 + 2v'_2
\Rightarrow 2v_1 + 15v'_1 + 36v'_2 = v'_1 + 2v'_2
\Rightarrow 2v_1 = -14v'_1 - 34v'_2
\Rightarrow v_1 = -7v'_1 - 17v'_2.
$$

$$
\boxed{
\begin{pmatrix}
v_1 \ v_2
\end{pmatrix}=
\begin{pmatrix}
-7 & -17 \\
5 & 12
\end{pmatrix}
\begin{pmatrix}
v'_1 \ v'_2
\end{pmatrix}.
}
$$

## Задание 3

В старом городе улицы устроены так, что они соединяют перекрёстки. Каждый перекрёсток соединён с другими при помощи мостовых улиц. Велосипедист хочет проехать по городу, проехав **по каждой улице ровно один раз**, и вернуться в исходную точку.

В городе **12 перекрёстков**. Все улицы — двусторонние. Всего проложено **21 улица** между различными перекрёстками. Из любого перекрёстка можно добраться до любого другого. Некоторые перекрёстки соединены несколькими улицами.

Ответьте на следующие вопросы:

* Может ли велосипедист совершить такую поездку?
* Если нет, то сколько улиц нужно добавить, чтобы это стало возможным?
* А если разрешено убирать улицы, то какое минимальное количество нужно убрать, чтобы поездка стала возможной?

### Решение 

В связном неориентированном мультиграфе (у вас допускаются несколько улиц между одними перекрёстками) существует Эйлеров цикл (проезд по каждой рёбру ровно один раз и возврат в исходную точку) тогда и только тогда, когда все вершины имеют чётную степень. $n=12$ перекрёстков, $E=21$ улица $\Rightarrow$ сумма степеней $\sum \deg v =2E=42$. Из этого следует, что число вершин нечётной степени чётно. Обозначим число нечётных вершин $2k$ (то есть $=0,1,2,\dots,6$).

**Может ли велосипедист совершить такую поездку?**

Наличие только числа вершин и числа рёбер не даёт однозначного ответа: граф может быть уже эйлеровым (все степени чётные, $k=0$) и тогда поездка возможна; а может иметь нечётные вершины и тогда - нет.

**Сколько улиц нужно добавить, чтобы это стало возможным?**

Если в графе $2k$ нечётных вершин, то достаточно добавить $k$ новых улиц, соединив нечётные вершины попарно (каждое добавление меняет чётность степеней двух вершин). Поскольку $2k\le12$, максимум $k=6$. Требуется ровно $k$ добавлений, где $k$ - половина числа нечётных вершин; а в худшем случае (если все $12$ вершин нечётные) - $6$ улиц.

**Если разрешено убирать улицы, какое минимальное количество нужно убрать, чтобы поездка стала возможной?**

Удаление одного существующего ребра также меняет чётность степеней двух его концов. Поэтому удаляя рёбра, можно уменьшать число нечётных вершин на $2$ за одно удаление только в том случае, когда удаляемое ребро соединяет две нечётные вершины. В теоретическом минимуме, если удачно выбирать ребра, потребуется удалить также $k$ улиц (тот же аргумент: нужно сократить $2k$ нечётных до $0$ - по два за удалённое ребро). В общем случае (если нужные ребра между нечётными вершинами отсутствуют) может потребоваться больше удалений, но минимально можно обойтись $k$ удалениями, и в худшем случае (при $2k=12$) это опять $6$ улиц.

## Задание 4

Пусть для дерева $ T $ функция $ D(T) $ задаёт сумму расстояний между всеми парами различных вершин $ u, v $:

$$
D(T) = \sum_{u,v \in V(T), , u \neq v} d(u, v)
$$

Какое дерево с $ n $ вершинами минимизирует эту функцию?
Докажите это и найдите формулу для $ D(T) $ в этом случае.

### Решение 

Для любого ребра $e$ дерева, при его удалении дерево распадается на два компонента с числами вершин $s_e$ и $n-s_e$. Сколько упорядоченных пар $(u,v)$ имеют путь, проходящий через $e$? Это все пары, у которых один конец в одном компоненте, другой - в другом, то есть $2\cdot s_e (n-s_e)$ (умножили на $2$, потому что считаем упорядоченные пары). Каждая такая пара даёт вклад $1$ в длину пути за счёт этого ребра. Следовательно, суммарно по всем рёбрам

$$
D(T)=\sum_{e\in E(T)} 2\cdot s_e (n-s_e).
$$

Рассмотрим функцию $f(s)=s(n-s)=ns-s^2$ на целых $s=1,\dots,n-1$. Это парабола с вершиной в $s=\tfrac n2$; её минимальные значения на отрезке принимаются на концах, т. е. при $s=1$ или $s=n-1$, и равно $f(1)=n-1$.

Отсюда для любого ребра $e$ имеем $s_e(n-s_e)\ge n-1$, причём равенство выполняется тогда и только тогда, когда $s_e=1$ или $n-1$. Значит $D(T)=\sum_{e} 2,s_e(n-s_e)\ge \sum_{e} 2(n-1)=2(n-1)^2$ поскольку в дереве ровно $n-1$ рёбер.

Равенство возможно тогда и только тогда, когда для каждого ребра $s_e=1$ или $n-1$, то есть каждое ребро отделяет ровно одну вершину от остальных. Это центральная вершина соединена рёбрами со всеми остальными.
