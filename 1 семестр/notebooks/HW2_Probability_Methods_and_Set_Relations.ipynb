## Задание 3

В научном кружке работают 72 профессора. За год каждый профессор может опубликовать статью в следующих типах журналов (независимо друг от друга):

Зарубежные журналы — с вероятностью 1/12.
Российские журналы — с вероятностью 1/4.
Онлайн-журналы — с вероятностью 1/6.
Вероятности независимы для каждого типа публикации.

Что нужно сделать:

Найдите вероятность, что ровно 4 профессора опубликовали статьи только в зарубежных журналах (то есть в зарубежных, но не в российских и не в онлайн-журналах).
Найдите вероятность, что ровно 5 профессоров опубликовали статьи в двух или трех категориях (то есть в любых двух или в трех).
Найдите вероятность, что ровно 6 профессоров не опубликовали статьи ни в одной из трех категорий.

### Решение 

```python
import math
```

Пусть:

$A$ - профессор опубликовал статью в зарубежном журнале: $p_A=\frac{1}{12}$.

$B$ - профессор опубликовал статью в российском журнале: $p_B=\frac{1}{4}$.

$C$ - профессор опубликовал статью в онлайн журнале: $p_C=\frac{1}{6}$.

#### 1.1 $A \land \neg B \land \neg C$ (только зарубежные)

Вероятность для одного профессора:

$$p_{только \space A} = p_A(1 - p_B)(1 - p_C)=\frac{1}{12} \cdot \frac{3}{4} \cdot \frac{5}{6} = \frac{5}{96}$$

Если профессоров $n = 72$ и мы ищем вероятность где ровно $k = 4$ таких, то используем биномы (события независимы).

$$
P_k = \binom{n}{k} p_{только \space A}^k \left(1 - p_{только \space A} \right)^{n-k}
$$

$$
P_k = \binom{72}{4} \left(\frac{5}{96}\right)^{4} \left(1 - \frac{5}{96} \right)^{68}
$$

```python
# параметры
n = 72
k = 4
p = 5/96

# биномиальная вероятность
P_k = math.comb(n, k) * (p**k) * ((1 - p)**(n - k))

print(P_k)
```

#### 1.2 Две или три категории

Вероятность для одного профессора:

2 категории: $(\neg A \land B \land C) + (A \land \neg B \land C) + (A \land B \land \neg C) = \frac{11}{288} + \frac{3}{288} + \frac{5}{288} = \frac{19}{288}$

3 категории: $(A \land B \land C) = \frac{1}{n*k} = \frac{1}{288}$

2 или 3 категории: $\frac{1}{288} + \frac{19}{288} = \frac{20}{288} = \frac{5}{72}$

Пользуясь ранее полученными результатами:

$$P_k=\binom{72}{5} \left(\frac{5}{72}\right)^{5} \left(1 - \frac{5}{72} \right)^{67}$$

```python
# параметры
n = 72
k = 5
p = 5/72

# биномиальная вероятность
P_k = math.comb(n, k) * (p**k) * ((1 - p)**(n - k))

print(P_k)
```

#### 1.3 $\neg A \land \neg B \land \neg C$ (ни в одном)

Вероятность для одного профессора:

$$p_{ни \space в \space одном}=(1 - p_A)(1 - p_B)(1 - p_C)=\frac{11}{12} \cdot \frac{3}{4} \cdot \frac{5}{6} = \frac{55}{96}$$

Пользуясь ранее полученными результатами:

$$P_k=\binom{72}{6} \left(\frac{55}{96}\right)^{6} \left(1 - \frac{55}{96} \right)^{66}$$

```python
# параметры
n = 72
k = 6
p = 55/96

# биномиальная вероятность
P_k = math.comb(n, k) * (p**k) * ((1 - p)**(n - k))

print(P_k)
```
