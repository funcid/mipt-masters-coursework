# Семинар 3. Декомпозиция метрик и основные фреймворки работы с метриками

# Метрики активностей пользователей

* DAU (Daily Active Users)
* WAU (Weekly Active Users)
* MAU (Monthly Active users)

# Абсолютные метрики (ванильные)

* Количество установок или регистраций, подписок;
* Количество скачиваний, зарегистрированных метрик

# Прокси-метрики

Это метрики-посредники, которые могут подсказать что-то о потенциальном успехе вашего продукта. Но поскольку они сами по себе не представляют прямого доказательства того, как ваш продукт будет взаимодействовать с пользователями, нужно быть осторожными с этими метриками.

* Sticky factor = DAU / MAU (прилипаемость)
* CAR (Conversion Rate) - конверсия
* Acquisition - привлечение
* Retention - возвращаемость
* Churn rate - показатель оттока пользователей

# Метрики, ориентированные на бизнес

* LTV (Lifetime value) - пожизненная стоимость пользователя, прогнозируемая
* AOV (Average order value) - средний чек
* ARPU / ARPPU (Average revenue per [paying] user) - средняя выручка на одного [платяжего] пользователя

# Что такое хорошая метрика?

1. Чувствительная к изменениям в продукте. Метрика должна реагировать на изменения в продукте как в положительную, так и в отрицательную сторону.
2. Понятная (интерпритируемая) - всегда однозначно растет или падает при улучшениях или ухудшениях в продукте.
3. Сравнимая - поддается сравнению с собой или рынком. В идеале мы можем сравнить значение метрики с конкурентами.
4. Реальная - можно посчитать метрику и управлять метрикой посредством продуктовых решений.
