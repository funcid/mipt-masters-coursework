### 1.1 Докажите, что для любого натурального числа $n$ верно  
$$1^3 + 2^3 + 3^3 + ... + n^3 = (1 + 2 + 3 + ... + n)^2$$

### Решение 

Решу задачу через метод математической индукции.

**Этап 1.** 
Проверим истинность при $n = 1: 1^3 = 1^2$; это верно

**Этап 2.** 
Предположим, что $1^3+2^3+3^3+...+n^3=(1+2+3+...+n)^2$, заметим, в правом выражении подойдет формула арифметической прогрессии. Нужно взять первый и последний элементы последовательности, умножить их на количество пар. В данном случае, $1+2+3+...+n =\frac{(n+1)n}{2}$, то есть 

$$1^3+2^3+3^3+...+n^3=\left(\frac{(n+1)n}{2}\right)^2$$

**Этап 3.** 
Сделаем шаг индукции $+1$: $$1^3+2^3+3^3+...+n^3+(n+1)^3=\left(\frac{(n+1+1)(n+1)}{2}\right)^2$$ 
В левой части уравнения, первые элементы образуют уже известное $\left(\frac{(n+1)n}{2}\right)^2$, поэтому перепишу уравнение 

$$\left(\frac{(n+1)n}{2}\right)^2+(n+1)^3=\left(\frac{(n+2)(n+1)}{2}\right)^2$$

Приведем к общему знаменателю:

$$\frac{(n+1)^2n^2+4(n+1)^3}{4}=\frac{(n+2)^2(n+1)^2}{4}$$

Домножим обе части уравнения на $4$:

$$(n+1)^2n^2+4(n+1)^3=(n+2)^2(n+1)^2$$

Выделим из $(n+1)^3$ выражение $(n+1)$ и выделим $(n+1)^2$ из уравнения:

$$(n+1)^2n^2+4(n+1)^2(n+1)=(n+2)^2(n+1)^2$$
$$(n+1)^2(n^2+4(n+1))=(n+1)^2(n+2)^2$$

Раскроем скобки:

$$n^2+4(n+1)=(n+2)^2$$
$$n^2+4n+4=n^2+4n+4$$
$$0=0$$

База верна, переход непротиворечив.

Разобьем выражение на последовательные части:

|0|1|1| 1 | 0 | 1 |

8. Сравните эффективность алгоритмов двоичного поиска, поиска в глубину и поиска в ширину с помощью O-нотации.


```python
def binary_search(arr, target):
    """
    Бинарный поиск в отсортированном массиве.
    Возвращает индекс элемента или -1, если не найден.
    """
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1


def search(graph, start, get_next):
    """
    Универсальный алгоритм обхода графа.
    graph: dict[node, list[neighbors]]
    start: стартовая вершина
    visit: функция, вызываемая при посещении вершины
    get_next: стратегия выбора следующего элемента из очереди
    """
    visited = set()
    queue = [start]

    while queue:
        node = get_next(queue) # pop или pop(0)
        if node not in visited:
            visited.add(node)
            # добавляем соседей
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    queue.append(neighbor)


def dfs(graph, start):
    return search(graph, start, lambda q: q.pop()) # LIFO


def bfs(graph, start):
    return search(graph, start, lambda q: q.pop(0)) # FIFO


# Пример
g = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(dfs(g, 'A'))
print(bfs(g, 'A'))
print(binary_search([1, 3, 5, 7, 9], 7))  # 3
```
