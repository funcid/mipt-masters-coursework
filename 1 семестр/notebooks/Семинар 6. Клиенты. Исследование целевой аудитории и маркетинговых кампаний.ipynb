# Семинар 6. Клиенты. Исследование целевой аудитории и маркетинговых кампаний

#### Продуктовая матрица

Продуктовая матрица - последвательность действий, которые знакомят потенциального клиента с вашим продуктом (услугой) и ведут к совершению новых покупок.

Логика последовательных предложений, которые ведут клиента от первого контакта к повторным покупкам.

#### Расширенная продуктовая матрица: основные элементы

**Задача:** провести клиента от первого контакта к повторным покупкам и подписке, растя LTV при контроле САС.

**Ядро матрицы из 5 элементов:** 
* Lead-magnet - привлечение клиента и добавление его в нашу базу. Это может быть бесплатный продукт, который не несет прибыли, но позволяет расширить базу клиентов.
* Tripwire - входной продукт, также не дает значительного прироста в экономическом эффекте, но позволяет закрепить и усилить зарождающиеся отношения с клиентами.
* Core offer - ключевой предоставляемый продукт. До этого рассматриваются входные, привлекающие, вспомогательные продукты. Вокруг Core offer строится весь бизнес.
* Profit maximization - сопутствующие товары. Механики, увеливающие маржу и повторы.
* Recurrent product - повторные покупки. Мы хотим иметь с клиентами долгосрочные отношения.

##### Дополнительные элементы (необязательно):
* One-time offer / Flash sale - уникальное срочное предложение, монетезация горячего интереса.
* One-click upsell - быстрая дополнительная покупка, быстрый рост AOV.
* VIP product - премиум-чек для "верха" спроса.
* Bundles - пакетные продажи. Специальная акция на группу товаров.
* Components - расходники/комплектующие для Core.
* Subscription - подписки и абонементы.

##### Контрольные метрики по этапам:
* Переходы: CR между ступенями, время до следующего шага.
* Доходность: AOV, маржа, CPA по каждому элементу.
* Удержание: Retention. Churn, LTV/CAC.
* Качество: NPS, Refund rate по Core.

#### **Задача:** распишите основные элементы продуктовой матрицы.

**Контекст:** онлайн-курс по аналитике данных для начинающих.

**Цель:** провести от первого контакта к подписке/повторам, увеличивая AOV и LTV.

Придумайте по одному варианту для: Lead-magnet, Tripwire, Core offer, Recurrenct product. Добавьте минимум два "училителя".


**Core-элементы:**
* Lead-magnet: вводный урок (собрать контакты, KPI: CR визит > лид)
* Tripwire: бесплатный урок на выбор (закрепить контакты, отсеять немотивированных, KPI: CR)
* Core offer: платный курс (доход с продажи, KPI: CR лид > покупка, AOV, маржа)
* Recurrenct product: доступ к сезонным урокам, другие курсы (проверка качества, удержание, KPI: NPS, LTV/CAC, Retention)


**Усилители (Profit maximization + Retention):**
* Пакет персональных доразборов ДЗ (One-click upsell, KPI: Upsell rate, триггер: студент сдал 2 ДЗ)
* Скидка на курс из другой рекомендуемой сферы (One-time offer, KPI: Retention, LTV, триггер: студент завершил 2 разных курса)
