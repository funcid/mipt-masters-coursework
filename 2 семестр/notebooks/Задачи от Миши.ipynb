## Задача 1

**Условие:** ты пришёл в казино у тебя на столе лежит кубик с 20 гранями пронумерованными от 1 до 20 (на нём уже какое-то число сверху). Игра в 100 раундов. Каждый раунд ты можешь либо взять кол-во долларов равное текущему числу на кубике, либо потратить раунд на то чтобы перебросить кубик можно брать деньги сколько угодно раундов подряд, перебрасывать тоже. Какая оптимальная стратегия и какой при ней ожидаемый выигрыш? 

**Решение:** у нас конечное число ходов, а значит можно применить метод обратных рассуждений и смотреть на ситуацию когда остался последний ход. Для начала построю модель Маркова. Пусть $S$ это количество оставшихся ходов, а $g$ это текущее состояние выигрыша. Тогда текущее состояние это $(S, g)$, а следующее зависит от детерминированного выбора.

1. Если я принимаю решение остановиться, то $(S, g) \rightarrow (S - 1, g)$, тогда ожидаемый выигрыш $S \times g$
2. Если я принимаю решение перебрасывать кубик, то $(S, g) \rightarrow (S - 1, \mathcal{X})$, где $\mathcal{X}$ это результат события при равномерном распределении от 1 до 20. , тогда $M(R(S - 1, \mathcal{X}))$

Спрашивается, а зачем тут вообще модель Маркова если вероятности не "перетекают" друг в друга? Во время тренировки эта идея казалось лучше...

Так как, делаем оптимальную модель, то один выигрыш должен быть больше второго:

$$S \times g > M(R(S - 1, \mathcal{X}))$$, так как при принятия решения $S >= 1$, то можно разделить обе части

$$g > \frac{M(R(S - 1, \mathcal{X}))}{S}$$, то есть есть такое $g$, которое нам подходит

$$g > \frac{1}{S} \frac{1}{20} \sum^{20}_{x=1}{R(S-1, x)}$$, как видно, $S$ имеет очень высокий вес и тренд функции почти не зависит от количества храний или значения $x$

Рассмотрим ситуацию, когда остался последний ход, следовательно $S = 1$ (база индукции):

$$g > \frac{1}{20} \sum^{20}_{x=1}{R(S-1, x)}$$

$$g > \frac{1}{20} \sum^{20}_{x=1}{x}$$

$$g > 10.5$$

Теперь у нас есть текущий бросок и возможность (если результат не понравится) перейти к состоянию $S = 1$, ценность которого мы уже знаем. То есть на предпоследнем ходе оставляем если $g > 10.5$, и перебрасываем в другом случае (шаг индукции).

$$g > \frac{10}{20} \times 10.5 + \frac{10}{20} \frac{11 + 20}{2}$$

$$g > 5.25 + 7.25$$

$$g > 12.5$$

Как я вижу, необходимый $g$ с конца растет довольно быстро (на $2$ за последний ход), учитывая что придется рекурсивно считать $100$ ходов, я сдаюсь и доверяю это машине, прости Миша, на глазок это 16-17, как свести рекурсию к точному числу без вычислений я хз

```python
# База индукции: за 0 раундов выигрыш 0
R = 0
S_total = 100

for S in range(1, S_total + 1):
    R = sum(max(S * x, R) for x in range(1, 21)) / 20

print(f"Ожидаемый выигрыш за 100 раундов: {R}")
print(f"Порог для последнего броска (g >): {R / S_total}")
```
