### Phrases

**Out loud** - вслух

**I work as software engineer** - я работаю разработчиком ПО

**Do you often go abroad?** - Вы часто ездите заграницу?

**To go on a business trip** - ездить в командировку

### Telephoning

**Can I speak to Peter, please?** - Могу ли я говорить с Питером, пожалуйста?

**Speaking** - Это я (Слушаю)

**Are you busy now?** - Вы сейчас заняты?

**Yes, actually, I'm a bit busy now. Can you call me back?** - Да, вообще-то я немного занят сейчас. Вы моежете мне перезвонить?

**Sure. When cal I call you back?** - Конечно. Когда я могу вам перезвонить?

**Can you talk now?** - Вы можете сейчас говорить?

**Yes, I can. Who is speaking?** - Да, могу. Кто говорит?

**This is a very bad line** - Связь очень плохая.

**You're breaking up** - Вы прерываетесь (пропадаете).

**I'm calling regarding...** - Я звоню насчет...

**Please, hold on** - Пожалуйста, оставайтесь на линии.

**Sorry to keep you waiting** - Извините, что заставили ждать

**I'll connect you** - Я соединю вас
