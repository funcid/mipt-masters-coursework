### Phrases

**I'm calling regarding our arrangements** - Я звоню насчет наших договоренностей

**I'll get him** - Я сейчас его позову

**It will be quicker to call you than** - Будет быстрее вам позвонить, чем...

**I'd like to run through our arrangements** - Я бы хотел пробежаться по нашим договоренностям

**Sorry to keep you waiting** - Извините, что заставил ждать
