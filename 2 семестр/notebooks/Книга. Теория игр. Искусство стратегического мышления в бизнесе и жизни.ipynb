# Книга. Теория игр. Искусство стратегического мышления в бизнесе и жизни

### Десять историй о стратегии

#### 1. Игра на угадывание числа

Нужно выбрать число в интервале [1; 100], конечно можно использовать инженерный подход (бинарный поиск, снижение энтропии), но вероятностное распределение не равномерное и зависит от человека, награда по формуле $1000р. - 200р. \times \space\text{кол-во ходов}$.

#### 2. Победа ценой поражения

Порой, испытание не одно и состоит из цепочки связных выборов. Если финальная цель победить, не обязательно выигрывать на каждом этапе.

#### 3. Счастливая рука

Сложно оценить результативность отдельного игрока, особенно в командном спорте. Когда "счастливая рука" начинает выигрывать, соперники концентрируют внимание на ней, не уделяя всего внимания другим соперникам. 

#### 4. Быть или не быть лидером

Зачастую, как в парусной регате, чтобы оставаться лидером, нужно копировать действия 2-ого места, однако это не работает если участников больше. В бизнесе, часто не один получает все, поэтому стратегия подражания очень выгодна.

#### 5. На том стою 

Стратегия "На том стою, ни шага назад", вынуждает оппоненотов продавливаться и выбирать не всегда выгодные варианты, однако это может быть дорого в долгосрочной перспективе и быть предсказуемой стратегией.

#### 6. Стратегическое похудения

Искуссетвенное ограничение кол-ва вариантов чтобы промотивировать или создать стимулы для действий. Например разрушение кораблей Кортеса чтобы у испанцев не было вариантов отступать от индийцев. 

#### 7. Дилемма Баффета

Рациональные игроки не всегда будут сотрудничать друг с другом, даже если это в их интересах. Если оба мыслят в личных интересах, то правильным решением будет только предательство. Лучше в такие истории просто не в вязываться.

#### 8. Смешивание ходов

Когда игра одна, выбор стратегии может мало на что повлиять, но если есть хоть шанс нащупать поведение игрока, то можно будет его переигрывать, чтобы защититься от этого, можно использовать рандомизированную стратегию, например для выборки фирм на налоговую проверку из всех.

#### 9. Пари для простаков

Если вам предлагают пари, скорее всего уже знают ваш выбор, поэтому при отсутствии информации лучше пари отвергать.

#### 10. Теория игр может быть опасной для здоровья 

Нужно понимать сторону вашего оппонента.

### Решение игр методом обратных рассуждений

#### Два типа стратегического взаимодействия 

Последовательный, параллельный (прямо как процессы). В последовательной игре, каждый **игрок должен проанализировать будущие ответные действия других игроков**. Вето и выборочное вето могут дать влияние одному из участников, однако большая свобода может уменьшить конечную сумму игры (выборочное вето).

**Задача:** вы выигрываете, если заставите другую команду взять последний флажок. Сейчас ваш ход и у вас 21 флажок. Сколько флажков вы возьмете? Можно брать [1; 3] флажка за ход 

**Рассуждения:** тот кто оставляет [2; 4] флажка - проиграл, так как можно свести к победе $2-1; 3-2; 4-3$. Соответсвенно, тот кто с 5-ью флажками всегда проигрывает. По той же логике, из любого 6,7,8 можно сделать 5. Это повторяется для $9, 13, 17, 21$. То есть первый игрок находится в проигрышной позиции при оптимальной игре соперника.

Важный принцип. В начале сессионной игры всегда нужно рисковать больше чем в конце. Это касается как и тенниса, так и инвестиций. В начале оптимизируется рост, в конце выживаение - ассиметрия последствий.

### Решение дилеммы заключенных

Основная идея заключается в том, что у игроков есть выбор предать или сотрудничать. В случае предательства одного, он получает выгоду, другой нет. Если оба сотрудничают - оба получают выгоду, меньшую чем в случае с предательством. Если оба мыслят в своих интересах - они предадут друг друга и это самый невыгодный вариант для обоих. 

Возможное решение заключается в вознаграждении участников со стороны третьего игрока, чтобы компенсировать риск предательства и оно стало невыгодным.

Но в реальной жизни, чаще используется наказание для предотвращения предательства. Каким требованиям должно отвечать наказание: обман должен быть определен, наказание должно быть определено, игроки должны понимать когда совершают обман, неизбежность скорого наказания, наказание должно быть соразмерным обману, повторяемость. 

## Прекрасное равновесие

### Равновесие Нэша

Равновесие Нэша в игре с параллельными ходами - это такое состояние стратегии для всех игроков, при котором никто не может улучшить свой результат, изменив свою стратегию в одиночку, если стратегии остальных игроков остаются неизменными.

### Фокальная точка

Фокальная точка - это решение, которое люди интуитивно выбирают по умолчанию, чтобы скоординировать действия без прямого общения.

## Выбор и случай

### Смешивание стратегий

В игре с нулевой суммой (чистый конфликт), стоит выбирать смешанную стратегию (несколько стратегий, выбор которой зависит от оптимальной пропорции) чтобы противник не мог воспользоваться своей стратегией.

На деле, данная стратегий противоречива, теоретически она удобна, практически трудно ей следовать, человек всегда действует по системе. Например в камень-ножницы-бумага, лучше использовать стратегию бумага-ножницы-бумага исключая третий вариант чтобы соперник думал о нем. 

## Стратегические ходы

**Безусловный первый ход** - обязательство, ставит другого игрока перед фактом, на который он должен отреагировать

**Правило ответной реакции, закрепляющее условный второй ход** - угроза/обещание, ответные ходы, наносят ущерб или вознаграждает игрока

### Предупроеждения и заверения

Угрозы и обещания принуждают делать то, что делать невыгодно, однако предупреждения и заверения носят информативный характер и никак не влияют на решения других игроков, потому что они выгодны для игрока и очевидны.

## Обеспечение достоверности стратегии

### Контракты

Контракт с третьей независимой стороной является важным инструментом обеспечения достоверности стратегии, так как позволяет привлечь внешнего участника для объективной оценки её реализации и результатов. В таком контракте фиксируются ключевые метрики, методы проверки и ответственность сторон, что повышает прозрачность процессов и снижает риск субъективной интерпретации данных. Привлечение независимой стороны способствует росту доверия со стороны заинтересованных участников и обеспечивает более надёжный контроль достижения поставленных целей.

### Репутация

Репутация является важным фактором обеспечения достоверности стратегии, так как отражает уровень доверия к организации со стороны клиентов, партнёров и других заинтересованных сторон. Положительная репутация формируется за счёт прозрачности, последовательности действий и выполнения обязательств, что служит косвенным подтверждением надежности реализуемой стратегии. В то же время репутационные риски стимулируют организацию придерживаться заявленных принципов и поддерживать высокий уровень качества принимаемых решений.

### Прекращение коммуникации

Прекращение коммуникации выступает как крайняя мера обеспечения достоверности стратегии, применяемая в случае утраты доверия к одной из сторон или выявления существенных нарушений договорённостей. Такой шаг позволяет минимизировать дальнейшие риски, связанные с искажением информации или недобросовестным поведением. При этом возможность разрыва взаимодействия заранее фиксируется в правилах или соглашениях, что дисциплинирует участников и способствует соблюдению установленных требований.

### Сжигание мостов

Сжигание мостов представляет собой радикальный механизм обеспечения достоверности стратегии, при котором исключается возможность возврата к прежним договорённостям или недобросовестным практикам. Такой подход демонстрирует серьёзность намерений и снижает риск отклонения от выбранного курса, поскольку делает изменения необратимыми. В результате повышается доверие к стратегии со стороны заинтересованных сторон, однако возрастает и цена ошибки, так как пространство для манёвра существенно ограничивается.

### Вывод результата из своей зоны контроля или его передача на волю случая

Вывод результата из своей зоны контроля или его передача на волю случая рассматривается как способ снижения возможности целенаправленного искажения итогов, поскольку участник теряет прямое влияние на результат. Это может реализовываться через автоматизированные механизмы, случайные процессы или передачу контроля внешней среде. Такой подход повышает объективность и доверие к результатам, однако одновременно снижает управляемость и увеличивает зависимость от внешних факторов.

### Поэтапное движение

Поэтапное движение предполагает реализацию стратегии через последовательные шаги с промежуточной проверкой результатов, что позволяет своевременно выявлять отклонения и корректировать действия. Такой подход повышает прозрачность процесса и снижает риски накопления ошибок.

### Командная работа

Командная работа обеспечивает распределение ответственности между участниками, что снижает вероятность субъективных решений и повышает объективность за счёт взаимного контроля. Совместная деятельность способствует более взвешенной реализации стратегии.

### Представители, уполномоченные вести переговоры

Представители, уполномоченные вести переговоры выступают как ответственные лица, действующие от имени сторон в рамках установленных полномочий. Это формализует процесс взаимодействия, снижает риск недопонимания и повышает надёжность достигнутых договорённостей.
