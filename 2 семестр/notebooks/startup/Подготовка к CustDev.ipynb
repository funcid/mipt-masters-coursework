
# Проблемное интервью и решенческое интервью

## 1. Фильтр-вопросы

### По сегменту
- Пользуетесь ли вы сейчас более чем одним банковским приложением или инвестиционной платформой (вклады / акции / облигации / ИИС / валюта / золото)?
- Как вы сейчас ведёте учёт и анализ своего инвестиционного портфеля (брокерские приложения, Excel/Google Sheets, специализированные сервисы)?

### Опыт решения проблемы/задачи
- За последние 6–12 месяцев вам приходилось регулярно отслеживать доходность и структуру всего портфеля целиком?
- Вы лично занимаетесь управлением и анализом своего инвестиционного портфеля?
- Сколько времени в среднем в месяц уходит на мониторинг и сверку данных по портфелю?
- Если нет — кто именно этим занимается?

---

## 2. Сегментация

### Задачи
- Какие задачи перед вами стоят при управлении и анализе инвестиционного портфеля?
- Когда в последний раз вам приходилось считать доходность или структуру портфеля вручную?
- В какой ситуации вам особенно важно видеть картину по всему портфелю целиком?
- Что именно вы тогда хотели понять или проверить?

### Мотивы
- Почему для вас важно регулярно анализировать инвестиционный портфель?
- Что для вас будет считаться «хорошим результатом» управления портфелем?
- Какая конечная цель управления портфелем для вас сейчас?
- Как вы понимаете, что двигаетесь к этой цели?

### Должности (для B2B)
- Где вы работаете?
- Какую должность занимаете?
- В каком подразделении?
- Чем занимается ваша компания?
- Совмещаете ли инвестиции с предпринимательской деятельностью?

### Специфика сегмента
- Сколько активов и брокерских счетов вы обычно учитываете одновременно?
- Как часто вы пересматриваете структуру портфеля или принимаете решения о ребалансировке?
- Что обычно становится триггером для таких решений?

---

## 3. Цели и мотивы сегмента

### Цели
- Почему вы регулярно занимаетесь анализом и мониторингом портфеля?
- Что для вас является конечным результатом этой работы?

### ЛПР / ЛВР
- Кто принимает окончательное решение об использовании или покупке сервисов для анализа инвестиций?
- Кто или что влияет на это решение (рекомендации, обзоры, опыт, цена)?

---

## 4. Проблема

### Проблемы сегмента
- С какими сложностями вы сталкиваетесь при управлении и анализе инвестиционного портфеля?
- Как часто возникает необходимость вручную сверять данные по разным счетам и активам?
- В каких ситуациях эти сложности проявляются сильнее всего?
- Сколько времени это обычно занимает?

### Причина проблем и мотивы
- Почему возникают эти сложности?
- Что могло бы помочь вам перейти в желаемое состояние управления портфелем?
- Что мешает получать полную и точную картину по портфелю?
- Чего сейчас больше всего не хватает для достижения этого результата?

---

## 5. Существующие решения

### Используемые решения
- Как вы сейчас решаете задачу анализа и мониторинга портфеля?
- Какие конкретные сервисы или инструменты используете?
- Насколько вы довольны текущим решением (1–10)?
- Достигли ли вы поставленных целей? Почему?

### Частота использования
- Как часто пользуетесь этими инструментами?
- В каких ситуациях обращаетесь к ним чаще всего?

### Проблемы текущих решений
- С какими сложностями сталкиваетесь?
- Какие функции или данные приходится получать вручную?

### Плюсы текущих решений
- Что вам нравится?
- Почему выбрали именно эти решения?

### Ресурсы
- Сколько времени и/или денег уходит на решение задачи текущим способом?
- Сколько теряете (в деньгах или возможностях), если проблема не решена?

### Эмоции
- Какие эмоции испытываете, сталкиваясь с проблемой?
- О чем думаете в этот момент?

---

## 6. Ожидаемый результат

### Идеальное решение
- Каким было бы идеальное решение?
- Как поймёте, что проблема решена полностью?
- Что должно измениться после решения?
- Опишите желаемое состояние управления портфелем.
- Какие задачи необходимо решить, чтобы проблема считалась закрытой?

### Ресурсы
- Чего сейчас не хватает для достижения результата?
- Какие ресурсы или инструменты могли бы помочь?

---

## 7. Эффекты

- Каких целей достигнете при идеальном решении?
- Как это повлияет на вас и ваши решения?
- Что изменится в жизни или инвестиционном поведении?
- Может ли это повлиять на ваше окружение?

---

## 8. Критерии выбора решения

- По каким характеристикам выбираете сервис?
- Какие характеристики критичны, а какие вторичны?
- Оцените важность характеристик (1–10).
- Какие характеристики неприемлемы при низком качестве?
- Что означает «хорошо реализованная» характеристика?
- По каким признакам понимаете, что она выполнена хорошо?

---

## 9. Каналы и влияние

### Каналы привлечения
- Где узнаёте о новых сервисах?
- В каких СМИ или сообществах обратили бы внимание?
- Какие источники вызывают доверие?
- Где ищете отзывы и сравнения?

### Агенты влияния
- Чьему мнению доверяете?
- Какие эксперты или блогеры могут повлиять?

### Процесс принятия решения (CJM)
- Опишите по шагам процесс выбора.
- Что становится триггером для поиска решения?

---

## 10. Портрет потребителя

- Расскажите о себе и своём инвестиционном опыте.
- Как давно инвестируете и какими инструментами пользуетесь?

# Решенческое интервью (продолжение)

- Какой интерфейс или способ представления информации был бы удобен?
- Какие дополнительные возможности сделали бы сервис ценным?
- Кого могли бы порекомендовать для разговора на эту тему?
- Можно ли пригласить вас на бета-тест или демонстрацию MVP?
- Какие три вещи сделали бы сервис «обязательным»?
- Что стало бы решающим фактором при выборе среди конкурентов?
