**2. Идея проекта реализуема (блокирующий критерий)**

По стоп-листу: проект близко заходит в “инвест-сервисы” и потенциально в зону инвестрекомендаций. В тексте есть “ИИ-подсказки/рекомендации по оптимизации портфеля” и мультиактивность с крипто. В текущей формулировке это может трактоваться как инвестиционное консультирование/рискованные рекомендации, что для курса является красной зоной. Как довести до “да”: убрать “рекомендации” и “оптимизацию” как совет, оставить “аналитику/алерты/контроль риска”, добавить дисклеймер “не является инвестрекомендацией”, зафиксировать, что сервис не прогнозирует цены и не говорит “покупай/продавай”.

0 / 10

**3. Изложен замысел продукта**

А. Что продается и какие задачи закрывает: да (дашборд, агрегация, мониторинг). Б. Проблема связана с решением через метрику: частично (время 3–5 часов/мес есть, но эффект и УТП не собраны вокруг одной метрики; “упущенная доходность 2–4%” висит отдельно). В. Острота обоснована и есть масштаб: частично/да (цифры и расчет 12 600 руб/мес есть, но логика “почему именно из-за разрозненности” и как вы это устраняете измеримо в продукте, прописана слабо). CustDev и покупатель: в целом да (инвесторы/ИП/аналитики), но сегменты слишком широкие, без стартового фокуса.

14 / 20

**4. Проведен теоретический анализ**

Источники 5–7+ есть, есть расчет масштаба (потери времени/стоимости). Не хватает четких 2–3 выводных тезисов “что именно доказывает теоанализ и какую цифру берем в гипотезы” одним блоком на слайде.

7 / 10

**5. Понятна предметная область реализации продукта (процессы)**

Процессы перечислены (мониторинг, сравнение, риск, налоги), но “какое изменение вносится в процесс” описано общо. Нужна связка до/после по одному сценарию: “как сейчас ведут мультиактивный портфель” → “как будет в FuncFund” → “какая метрика улучшается”.

13 / 20

**6. Конкурентное преимущество и инновационность**

Отличия сформулированы на уровне общих обещаний (все в одном месте, уведомления, макроиндикаторы). Нет явного конкурентного сравнения и 1–2 преимуществ, которые реально трудно повторить. “ИИ-уведомления/рекомендации” без конкретики и без доказуемого механизма выглядят как маркетинг.

9 / 20

**7. Технологическое решение**

Направления названы, но стек и реализуемость не раскрыты: какие именно интеграции (формат импортов, источники данных), как решаете нормализацию активов/идентификаторов, хранение, частоту обновления, качество данных, безопасность (особенно если учет брокерских данных), архитектуру.

6 / 10

##### Обший комментарий

Добрый день! Спасибо за работу. На данном этапе есть время "переобуться" и доработать идею. Надеюсь, мои рекомендации вам помогут!

Что доработать в первую очередь:

Переписать позиционирование: не “рекомендации по оптимизации”, а “аналитика структуры, риск-метрики, алерты о событиях и отклонениях”. Убрать любые формулировки buy/sell/hold, “повышаем доходность”, “оптимизируем портфель”.

Зафиксировать одну ключевую проблему и метрику: например “время на сверку/консолидацию портфеля” или “ошибки учета/пропуски событий”. “2–4% упущенной доходности” лучше убрать на этом этапе или очень аккуратно переосмыслить как “ошибки ребалансировки из-за отсутствия видимости”, но без обещаний.

Сузить MVP и сегмент: стартовый сегмент 1 (например, частные инвесторы с 2+ брокерами и MOEX+банковские инструменты), без крипто на 1 семестр.

УТП в одну фразу по конструкту курса: “агрегация + контроль” и измеримый эффект (время/ошибки), без “доходности”.

Конкуренты: 3–5 прямых/косвенных (портфельные трекеры, брокерские приложения, Excel/Google Sheets), сравнение по 3 параметрам и где ваше отличие.

Технологии: добавить 5–7 строк “как работает” (ETL/нормализация, расписание обновлений, модель данных активов, алерты), чтобы было видно, что это инженерно реализуемо.
