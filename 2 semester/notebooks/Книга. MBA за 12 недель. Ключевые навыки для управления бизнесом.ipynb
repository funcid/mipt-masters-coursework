# Часть I. Цифры

## Глава 1: Ценность

Цель бизнеса - создание ценности для всех заинтересованных лиц (stakeholders)

Ценность для акционеров - это дисконтированные будущие чистые денежные потоки компании 

Так как будущая прибыль может быть далеко, она имеет меньший вес в оценке ценности, поэтомы приходится занижать или дисконтировать цену

На что можем повлиять?

1. Рентабельность (доходы - затраты)
2. Рост (больше продаж, клиентов)
3. Риск (уменьшить коэф. дисконтирования)

## Глава 2: Рентабельность

Рентабельность - это разрыв между тем, сколько клиент готов заплатить за продукт, и стоимостью ресурсов, затраченных бизнесом на его производство

Инвесторы не знают, вернут ли они когда-нибудь свой капитал. Они ожидают компенсации за свои переживания

Отчет о прибылях и убытках:

1. Показывает продажи и понесенные расходы за определенный период, обычно квартал или год
2. Классифицирует расходы в зависимости от цели, которой они служат

## Глава 3: Рост

Рост прибыли - цель. А рост продаж, как правило, лишь средство ее достижения

У кредиторов больше прав к свободному обороту компании чем у инвесторов

Сколь бы ни были обоснованы данными наши прогнозы, это лишь прогнозы

Как гласит шутка, экономисты успешно предсказали девять из последних пяти рецессий

Рост нужно прогнозировать, а не отчитываться задним числом. Прогнозы могут сбыться, а могут и нет

## Глава 4: Риск

Компании дают обещания на будущее заинтересованным сторонам - обещания, как правило, выраженные в форме денежных потоков. Будущие денежные потоки всегда подвержены неопределенности.

## Глава 5: Баланс

Баланс - это запись прошлых транзакций. Он отображает и то, чем владеет или что контролирует компания (ее активы), и то, что она должна (ее обязательства и капитал)

Основное бухгалтерское уравнение (активы = обязательства + капитал) верно по определению. Если актив баланса не равен пассиву, значит где-то есть бухгалтерская ошибка

Баланс учитывает фактор времени, раличая текущие и долгосрочные обязательства и активы

Как отчет о финансовом положении баланс дает представление о том, какие возможности и угрозы ожидают компанию в будущем

## Глава 7: Денежный поток и рабочий капитал

Управление рабочим капиталом - это ключевой фактор создания ценности для акционеров

Когда мы платим поставщикам быстрее, чем наши клиенты платят нам, динамика денежных потоков может убить компанию

Положительный коммерческий рабочий капитал делает компанию уязвимой для рисков:

1. Потребность в больше капитале при росте компании
2. Угроза неплатежей клиентов
3. Угроза убытков из-за испрченных или устаревших запасов

Отрицательный рабочий капитал возникает, когда компании собирают деньги с клиентов быстрее, чем платят поставщикам, при этом поддерживают низкий уровень запасов

## Глава 8: Структура затрат

Установление цен ниже переменных затрат подобно активации кнопки самоуничтожения бизнеса. Это не означает, что компании так никогда не поступают

$$\text{Точка бызубыточноти} = \frac{(\text{Цена за единицу} - \text{переменные затраты на единицу})}{(\text{Постоянные затраты})}=\frac{\text{Маржинальная прибыль}}{\text{Постоянные затраты}}$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Исходные данные
price_per_unit = 100          # Цена за единицу (P)
variable_cost_per_unit = 60   # Переменные затраты на единицу (VC)
fixed_costs = 20000           # Постоянные затраты (FC)

# Маржинальная прибыль
margin_profit = price_per_unit - variable_cost_per_unit

# Точка безубыточности
break_even_point = fixed_costs / margin_profit
break_even_revenue = break_even_point * price_per_unit

print(f"Маржинальная прибыль: {margin_profit}")
print(f"Точка безубыточности: {break_even_point:.2f} единиц")

# Диапазон значений
units = np.arange(0, 1000)

# Линии графика
revenue = units * price_per_unit
variable_costs = units * variable_cost_per_unit
total_costs = fixed_costs + variable_costs

# Построение графика
plt.figure(figsize=(12, 7))

# Основные линии
plt.plot(units, revenue, label="Выручка (P × Q)", linewidth=2)
plt.plot(units, variable_costs, label="Переменные затраты (VC × Q)", linewidth=2)
plt.plot(units, total_costs, label="Общие затраты (FC + VC × Q)", linewidth=2)

# Постоянные затраты
plt.axhline(
    y=fixed_costs,
    linestyle="--",
    label=f"Постоянные затраты (FC = {fixed_costs})"
)

# Точка безубыточности
plt.scatter(
    break_even_point,
    break_even_revenue,
    s=120,
    zorder=5
)

# Подписи переменных
plt.annotate(
    f"""
Точка безубыточности

Q = {break_even_point:.0f}
FC = {fixed_costs}
P = {price_per_unit}
VC = {variable_cost_per_unit}
MP = {margin_profit}
""",
    xy=(break_even_point, break_even_revenue),
    xytext=(break_even_point + 80, break_even_revenue - 15000),
    arrowprops=dict(arrowstyle="->"),
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.5", alpha=0.2)
)

# Формула на графике
formula = (
    r"$Q = \frac{FC}{P - VC}$" "\n"
    r"$MP = P - VC$"
)

plt.text(
    50,
    max(revenue) * 0.75,
    formula,
    fontsize=14,
    bbox=dict(boxstyle="round", alpha=0.15)
)

# Оформление
plt.title("График точки безубыточности")
plt.xlabel("Количество продукции (Q)")
plt.ylabel("Деньги")
plt.legend()
plt.grid(True)

plt.xlim(0, 1000)
plt.ylim(0, max(revenue) * 1.05)

plt.show()
```

## Глава 9: Основы оценки

Внутренняя стоимость компании - это сумма ее ожидаемых будущих чистых денежных потоков, дисконтированных на стоимость капитала.

Стоимость капитала - это минимальная доходность, необходимая инвесторам в качестве компенсации за отказ от безрисковой ставки, которую они могли бы получить, вложив средства в безопасный актив, и за риск того, что денежные потоки не будут содействовать ожиданиям.

Чтобы расчитать внутреннюю стоимость, мы прогнозируем будущие чистые денежные потоки и дисконтируем каждый из них.

```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Параметры модели
# -----------------------------
initial_cash_flow = 10000   # Начальный денежный поток
growth_rate = 0.08          # Рост денежного потока (8%)
discount_rate = 0.12        # Ставка дисконтирования (12%)
years = 10                  # Количество лет

# -----------------------------
# Расчёт денежных потоков
# -----------------------------
future_cash_flows = []
discounted_cash_flows = []

for year in range(1, years + 1):
    # Прогнозируем будущий денежный поток
    cash_flow = initial_cash_flow * (1 + growth_rate) ** year
    future_cash_flows.append(cash_flow)

    # Дисконтируем его к текущей стоимости
    discounted_cf = cash_flow / (1 + discount_rate) ** year
    discounted_cash_flows.append(discounted_cf)

# Внутренняя стоимость
intrinsic_value = sum(discounted_cash_flows)

print(f"Внутренняя стоимость: {intrinsic_value:,.2f}")

# -----------------------------
# Визуализация
# -----------------------------
x = np.arange(1, years + 1)

plt.figure(figsize=(12, 7))

# Будущие денежные потоки
plt.plot(
    x,
    future_cash_flows,
    marker='o',
    linewidth=2,
    label='Будущие денежные потоки'
)

# Дисконтированные денежные потоки
plt.plot(
    x,
    discounted_cash_flows,
    marker='o',
    linewidth=2,
    label='Дисконтированные денежные потоки'
)

# Подписи для каждого года
for i in range(years):
    plt.annotate(
        f"{discounted_cash_flows[i]:.0f}",
        (x[i], discounted_cash_flows[i]),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center'
    )

# Формула DCF
formula = (
    r"$PV = \frac{CF_t}{(1+r)^t}$"
    "\n"
    r"$Intrinsic\ Value = \sum PV_t$"
)

plt.text(
    1,
    max(future_cash_flows) * 0.75,
    formula,
    fontsize=16,
    bbox=dict(boxstyle="round", alpha=0.15)
)

# Итоговая стоимость
plt.axhline(
    y=intrinsic_value,
    linestyle='--',
    label=f'Intrinsic Value = {intrinsic_value:,.0f}'
)

# Оформление
plt.title("DCF — Расчёт внутренней стоимости")
plt.xlabel("Год")
plt.ylabel("Денежный поток")
plt.legend()
plt.grid(True)

plt.show()
```
