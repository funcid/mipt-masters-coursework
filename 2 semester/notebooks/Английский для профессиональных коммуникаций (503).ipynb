## Theory

### PPF

Структура короткой самопрезентации, которая помогает быстро и логично рассказать о себе.

* **Present** — кто ты сейчас
  Текущая роль, специализация, чем занимаешься.

* **Past** — что привело тебя сюда
  Ключевой опыт, проекты, достижения, технологии.

* **Future** — куда хочешь двигаться
  Цели, мотивация, почему тебе интересна эта роль/компания.

Используется для ответов вроде:

> “Tell me about yourself”

### Technical Narrative

Инженерная структура ответа для рассказа о сложном техническом проекте.

Обычно строится так:

1. **Context** — контекст проекта
   Что это был за продукт/система, команда, цель.

2. **Problem** — техническая проблема
   Ограничения, bottleneck, масштабирование, reliability, latency и т.д.

3. **Solution** — как ты это решал
   Архитектура, стек, trade-offs, принятые решения, почему именно так.

4. **Result (Impact)** — итог
   Метрики, производительность, стабильность, бизнес-эффект.

Главный фокус — инженерное мышление и процесс принятия решений, а не просто перечисление технологий.

### STAR

* **Situation** — ситуация
  Контекст и проблема.

* **Task** — твоя задача/ответственность
  Что нужно было сделать именно тебе.

* **Action** — конкретные действия
  Что ты сделал, как принимал решения, как взаимодействовал с людьми.

* **Result** — результат
  Итог, выводы, измеримый эффект, чему научился.

Используется для вопросов про:

* конфликты,
* ошибки,
* лидерство,
* неопределённость,
* дедлайны,
* командную работу.

Главное правило STAR:
не уходить в абстракции и делать акцент на конкретных действиях и результате.

## Practice

**Tell me about yourself**

Hi, I’m Tsaryuk Artyom. I’ve been working as a software engineer at MTS Web Services for the last three years. Our team develops fintech solutions related to the digital ruble, including transfers, payments, and account operations for both business and retail customers.

I actually started my journey in game development. I built Minecraft plugins and mods using Java and Kotlin, and created several game projects with more than 1,000 daily active users. Later, I joined SberTech, where I worked on a search engine platform.

Now I’d like to deepen my fintech expertise, work on larger-scale distributed systems, and eventually grow into a technical lead role. That’s why this opportunity is especially interesting to me.

**Tell me about the most technically challenging project you’ve worked on**

I’d like to talk about a QR payment project related to the digital ruble platform.

The system involved several actors: a mobile banking customer, a POS terminal used by a business client, and the central bank platform responsible for digital ruble operations. My responsibility was to implement QR-code-based payments from retail customers to merchants.

The main challenge was to combine strict security and compliance requirements with high scalability and reliability. We also had to support asynchronous interactions between multiple external systems while keeping transaction processing consistent.

To solve this, I proposed a microservice architecture with an event-driven approach. I developed the QR service and part of the payment-processing logic, integrated them with our internal platform and the central bank APIs, and designed asynchronous communication between services using messaging infrastructure.

After testing and deployment, we successfully launched the feature and enabled digital-ruble QR payments for our customers.
