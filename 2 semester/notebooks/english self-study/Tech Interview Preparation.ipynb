## Interview Structure & Preparation Notes

This interview is mostly discussion-driven, built around one practical scenario.
You start with a small coding task, and then the interviewer uses it to walk through databases, transactions, architecture, and production concerns across the full software lifecycle.

#### High-level structure

1. Coding exercise (≈ 20–25 min)
2. Databases & transactions (≈ 20–25 min) — the hardest part
3. Architecture, microservices, production (≈ 15 min)

### 1. Coding Exercise — Money Transfer Service

#### Purpose

They are looking for: clean domain thinking, attention to detail, awareness of concurrency and correctness</br>

#### Task

Implement a simple money transfer service (possibly pseudocode).</br>

#### What they pay attention to

Domain modeling; Use a proper Money type; Clear separation of responsibilities (service, domain, validation); Validation (very important)</br>

1. Transfer amount must be positive
2. Source and destination accounts must be different
3. Sender must have enough balance

Exceptions

1. Explicit exception handling
2. Ideally domain-specific exceptions
3. InsufficientFundsException
4. InvalidAmountException
5. SameAccountTransferException

Concurrency

Be aware of race conditions and deadlocks</br>
Simple approaches are fine (synchronized, locks)</br>
Higher-performance solutions are a plus, not a requirement</br>

Red flags

1. Money as float
2. No validation
3. Basic language mistakes (== vs .equals(), etc.)

### 2. Databases & Transactions (Most Important Section)

#### Core topics

##### ACID

What ACID means in practice

Where and why violations might occur

How it applies to money transfers

##### Locking

Optimistic vs pessimistic locking

Differences and trade-offs

When you would choose one over the other

Typical scenario:

Two users update the same bank account balance at the same time — how do you handle it?

##### Isolation levels

Read Uncommitted; Read Committed; Repeatable Read; Serializable

You should be able to explain: what problems each level solves; which level fits the scenario and why

##### SQL & practical details

How to lock a row in SQL (SELECT ... FOR UPDATE)

Basic query optimization for performance

##### Indexes

B-tree; Hash; Pros and cons

When indexes help and when they hurt

##### Scaling the database

Money transfer in a distributed or large-scale setup:

Distributed transactions; Externalized locking; CAP theorem

Also be familiar with:

Sharding; Replication; Partitioning

Differences, trade-offs, and use cases

##### Why people fail here

“Spring handles transactions for me”

Can’t explain why a certain approach is chosen

Weak connection between theory and real behavior

### 3. Architecture, Microservices & Production

#### Microservices architecture

Main concepts:

CQRS, Domain-Driven Design (DDD), Event Sourcing

They expect: what these are; why you would use them; consequences and trade-offs (complexity, consistency, maintenance)

#### Production environment

Stability patterns: Circuit breakers; Timeouts; Retries; Fallbacks

Usually they ask you to explain one or two.

Observability

Typical answer includes: logs; metrics; distributed tracing; business metrics

Example question:

What observability setup would you propose for this system?

#### Deployment & release strategies

Blue–Green deployments; Canary releases; Rollback strategies

Also:

example migration strategy; how to recover if a deployment fails
