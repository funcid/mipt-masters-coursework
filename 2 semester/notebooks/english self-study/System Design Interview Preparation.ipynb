# System Design Interview — Key Notes

https://www.tryexponent.com/blog/system-design-interview-guide

Конспект и структура подготовки к system design интервью (1 час).

## Общая идея
- Интервью субъективное, сильно оценивается **мышление и коммуникация**.
- Нужно спроектировать систему **с нуля**.
- Важно объяснять решения так, будто обучаешь junior-инженера.
- Всегда делать **осознанные выборы**, а не спрашивать «а как лучше?».

## Типы задач
- Lifestyle-фичи: бронирование отелей, билетов, поездов
- Card distribution: выпуск, доставка, активация карты
- ATM системы: backend-операции, взаимодействие с картой

## Структура интервью (5 шагов)
### Тайминг (пример)
- Step 1: Анализ — ~10 мин
- Step 2–3: Дизайн — ~30 мин
- Step 4: Рефайнмент — ~15 мин
- Step 5: Wrap-up — ~5 мин

## Step 1 — Analysis (обязательно)
❗ Пропустил = провалил

Что делать:
- Сбор **functional / non-functional** требований
- Уточнение границ и ограничений
- Задавать **уточняющие вопросы**
- Не делать предположений без подтверждения

Зачем:
- Меньше ошибок дальше
- Показывает бизнес-мышление

## Step 2 — High Level Design
**Скелет системы**:
- Компоненты (DB, Queue, Services)
- Взаимодействия компонентов
- End-to-end flow

Важно:
- Проговаривать вслух все решения
- Предлагать **варианты** и **выбирать один**
- Чёткие роли компонентов
- Можно указать sync/async, actors

## Step 3 — Low Level Design
Обязательное:
- API (эндпоинты)
- Data schema
- Message content
- Authentication & Encryption

Дополнительно (если есть время):
- Network (VPC, VPN)
- Security bootstrap

## Step 4 — Refinement
Цель — сделать систему **реально работоспособной**.

Критично:
1. Failure scenarios (в т.ч. бизнесовые)
   - падение провайдера
   - внешние зависимости
2. Monitoring инфраструктуры
3. Configuration updates без downtime

Дополнительно:
- Backup-провайдеры
- Idempotency
- Maintenance / bootstrap процедуры
- Tracking бизнес-процессов

## Масштабирование
Минимум:
- Рост пользователей без деградации

Идеал:
- Global scale
- Regional deployments
- Без центральной точки отказа

## Step 5 — Wrap-up
Что делать:
- Вернуться к требованиям
- Показать, как дизайн их покрывает
- Связать все решения между собой

Зачем:
- Убирает сомнения интервьюера
- Улучшает восприятие коммуникации
- Возможность предложить улучшения
- Демонстрация зрелости мышления
